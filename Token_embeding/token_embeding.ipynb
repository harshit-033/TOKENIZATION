# ***import pretrained model***

```python
import gensim.downloader as api
model=api.load("word2vec-google-news-300")
```

```python
vector1=model
```

```python
print(vector1["person"])
```

```python
len(vector1["person"])
```

```python
len(vector1)
```

## similar words

```python
print(vector1.most_similar("person"))
```

### king + women - men = ?

```python
print(vector1.most_similar(positive=["king","women"],negative=["man"]))
```

```python
# similarity between words
print(vector1.similarity("men","women"))
print(vector1.similarity("men","car"))
print(vector1.similarity("men","boy"))
print(vector1.similarity("car","engine"))

# similarity between men and car is very less comapred to car<->engine and men<->human
```

```python
import numpy as np
word1 = 'man'
word2 = 'woman'

word3 = 'semiconductor'
word4 = 'earthworm'

word5 = 'nephew'
word6 = 'niece'
vector_difference1 = model[word1] - model[word2]
vector_difference2 = model[word3] - model[word4]
vector_difference3 = model[word5] - model[word6]
magnitude_of_difference1 = np.linalg.norm(vector_difference1)
magnitude_of_difference2 = np.linalg.norm(vector_difference2)
magnitude_of_difference3 = np.linalg.norm(vector_difference3)

print("The magnitude of the difference between '{}' and '{}' is {:.2f}".format(word1, word2, magnitude_of_difference1))
print("The magnitude of the difference between '{}' and '{}' is {:.2f}".format(word3, word4, magnitude_of_difference2))
print("The magnitude of the difference between '{}' and '{}' is {:.2f}".format(word5, word6, magnitude_of_difference3))
```

# Creating token embedings

```python
import torch
```

```python
input_ids=torch.tensor([2,3,5,1])
```

```python
vocab_size=6
output_dim=3
torch.manual_seed(123)
embeding_layer=torch.nn.Embedding(vocab_size,output_dim)
```

```python
print(embeding_layer)
```

```python
print(embeding_layer.weight)
```

```python
print(embeding_layer(torch.tensor([3])))
```

```python
print(embeding_layer(input_ids))
```
